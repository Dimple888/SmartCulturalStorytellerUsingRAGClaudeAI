## 2. Data Understanding & Preparation

### a. Dataset Source
- Source: Synthetic dataset created from Sundarakanda (Ramayana) and public domain cultural texts.
- Data includes: story chunks, emotion/character labels, scene descriptions.

### b. Data Loading and Exploration
- In this project, data loading and exploration are performed in Java (Spring Boot backend) for integration with the RAG and LLM pipeline.
- Example Java code for loading and exploring the dataset is shown below.

#### Java Example: Load and Explore Dataset
```java
// Load story chunks from JSON in Spring Boot
Resource resource = new ClassPathResource("cultural-texts/story-dataset.json");
ObjectMapper mapper = new ObjectMapper();
List<StoryChunk> chunks = mapper.readValue(resource.getInputStream(), new TypeReference<List<StoryChunk>>(){});
System.out.println("Total Chunks: " + chunks.size());
System.out.println("Sample Chunk: " + chunks.get(0));
```

## 3. Model / System Design

### a. AI Technique Used
- Hybrid: Retrieval-Augmented Generation (RAG) + Large Language Model (LLM, Claude AI) + Multimedia (TTS, Anime)

### b. Architecture or Pipeline Explanation
- User query → RAG vector search → Story chunk selection → LLM-based story generation (Claude AI) → Scene extraction → Multimedia prompt generation (TTS, Anime)

### c. Justification of Design Choices
- RAG enables context-aware, relevant story generation.
- LLM (Claude AI) provides natural language generation and adaptation.
- Modular multimedia output (audio, anime) increases engagement and accessibility.
